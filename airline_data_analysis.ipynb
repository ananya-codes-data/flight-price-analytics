Does price vary with airlines?

```python
avg_price_airline = df.groupby('airline', as_index= False).agg(avg_price=('price', 'mean')).sort_values(by='avg_price', ascending= False).round(2)
print("Average price by airline:")
avg_price_airline
```

Does ticket price change based on departure time and arrival time?

```python
avg_dept_price = df.groupby('departure_time', as_index= False).agg(avg_price=('price', 'mean')).sort_values(by='avg_price', ascending= False).round(2)
print("Average departure price:")
avg_dept_price
```

```python
avg_arr_price = df.groupby('arrival_time', as_index= False).agg(avg_price=('price', 'mean')).sort_values(by='avg_price', ascending= False).round(2)
print("Average arrival price:")
avg_arr_price
```

How does price change with source and destination (route)?

```python
df['destination_city'].unique()
```

```python
df['source_city'].unique()
```

```python
df["route"] = df["source_city"] + " - " + df["destination_city"]
df.head()
```

```python
avg_route_price = df.groupby("route").agg(avg_price=('price', 'mean')).sort_values(by= 'avg_price', ascending=False)
avg_route_price
```
